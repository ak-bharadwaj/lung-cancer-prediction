# Exploratory Data Analysis & CatGBM Ensemble Benchmark
## Survey Lung Cancer Dataset Analysis

```python
from pathlib import Path

import pandas as pd

csv_path = Path('../data/raw/survey_lung_cancer.csv')
df = pd.read_csv(csv_path)
df.columns = df.columns.str.strip().str.replace(' ', '_').str.upper()
print('Dataset shape:', df.shape)
df.head()
```

```python
# 1. Missing Value and Duplicate Audit
print('Null values count:', df.isnull().sum().sum())
print('Duplicate rows count:', df.duplicated().sum())
print('Class distribution:\n', df['LUNG_CANCER'].value_counts(normalize=True) * 100)
```

```python
# 2. Demographic Statistics
print('Age Distribution by Target:')
df.groupby('LUNG_CANCER')['AGE'].describe()
```

```python
# 3. CatGBM Pipeline Evaluation
from ml.evaluate import evaluate_test_set

eval_results = evaluate_test_set()
print('Test Set Accuracy:', eval_results['metrics']['accuracy'])
print('Test Set Macro-F1:', eval_results['metrics']['macro_f1'])
print('Test Set ROC-AUC:', eval_results['metrics']['roc_auc'])
print('Test Minority (NO) Recall:', eval_results['metrics']['minority_class_no']['recall'])
```
